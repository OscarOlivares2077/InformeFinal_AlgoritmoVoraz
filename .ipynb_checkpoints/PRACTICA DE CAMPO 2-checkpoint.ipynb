```python
'''Caso 4: "Filtro de Spam en Log de Correos"
Contexto: Un servidor de correo registra 30,000 direcciones IP que intentaron enviar
mensajes.
• El Reto: Genera las IPs de forma aleatoria (puedes usar strings como "192.168.1.X").
• Desarrollo:
1. Implementa Quicksort para ordenar las IPs (debes tener cuidado con cómo
Python compara strings).
2. Una vez ordenadas, usa Búsqueda Binaria para verificar si 5 IPs específicas
(una "lista negra") están presentes en el log.
3. Reporta: ¿Cuántas comparaciones hizo la búsqueda binaria vs. la cantidad
total de registros?'''

import random

# Generar 30,000 IPs aleatorias
def generar_ips(n):
    return [f"{random.randint(1, 255)}.{random.randint(0, 255)}.{random.randint(0, 255)}.{random.randint(0, 255)}" for _ in range(n)]

# Función para convertir IP string a tupla de enteros para comparar correctamente
def ip_to_tuple(ip):
    return tuple(map(int, ip.split('.')))

ips_log = generar_ips(30000)
lista_negra = [ips_log[random.randint(0, 29999)] for _ in range(3)] + ["1.1.1.1", "2.2.2.2"]

def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivote = arr[len(arr) // 2]
    pivote_tuple = ip_to_tuple(pivote)
    
    izq = [x for x in arr if ip_to_tuple(x) < pivote_tuple]
    centro = [x for x in arr if ip_to_tuple(x) == pivote_tuple]
    der = [x for x in arr if ip_to_tuple(x) > pivote_tuple]
    
    return quicksort(izq) + centro + quicksort(der)

# Ordenar las IPs
ips_ordenadas = quicksort(ips_log)
def busqueda_binaria(arr, objetivo):
    bajo = 0
    alto = len(arr) - 1
    comparaciones = 0
    
    objetivo_tuple = ip_to_tuple(objetivo)
    
    while bajo <= alto:
        comparaciones += 1
        medio = (bajo + alto) // 2
        medio_tuple = ip_to_tuple(arr[medio])
        
        if medio_tuple == objetivo_tuple:
            return True, comparaciones
        elif medio_tuple < objetivo_tuple:
            bajo = medio + 1
        else:
            alto = medio - 1
            
    return False, comparaciones

print(f"{'IP a Buscar':<16} | {'Encontrada':<10} | {'Comparaciones'}")
print("-" * 45)

total_comps = 0
for ip in lista_negra:
    encontrada, comps = busqueda_binaria(ips_ordenadas, ip)
    total_comps += comps
    print(f"{ip:<16} | {str(encontrada):<10} | {comps}")

print("-" * 45)
print(f"Total de registros en el log: {len(ips_ordenadas)}")
print(f"Promedio de comparaciones por búsqueda: {total_comps / 5:.2f}")
    
```
