```python
#Montecarlo
#Imagina que tienes un cuadrado de lado 2 y dentro un círculo de radio 1. Si lanzas dardos al azar, la proporción de dardos que caen dentro del círculo nos permite calcular pi 
import random
import matplotlib.pyplot as plt

def estimar_pi(n_dardos):
    dentro_del_circulo = 0
    x_dentro, y_dentro = [], []
    x_fuera, y_fuera = [], []

    for _ in range(n_dardos):
        x, y = random.uniform(-1, 1), random.uniform(-1, 1)
        if x**2 + y**2 <= 1:
            dentro_del_circulo += 1
            x_dentro.append(x)
            y_dentro.append(y)
        else:
            x_fuera.append(x)
            y_fuera.append(y)

    pi_estimado = 4 * dentro_del_circulo / n_dardos
    
    # Visualización
    plt.figure(figsize=(6,6))
    plt.scatter(x_dentro, y_dentro, color='blue', s=1, label='Dentro')
    plt.scatter(x_fuera, y_fuera, color='red', s=1, label='Fuera')
    plt.title(f"Estimación de Pi con {n_dardos} dardos: {pi_estimado}")
    plt.legend()
    plt.show()

# Prueba con 5,000 dardos
estimar_pi(10)
```

```python
#Las vegas
#Aquí el objetivo es encontrar un juguete en un cuarto oscuro, eligiendo lugares al azar hasta que lo toquemos.
import random

def algoritmo_las_vegas(lista, objetivo):
    intentos = 0
    encontrado = False
    
    while not encontrado:
        intentos += 1
        indice_azar = random.randint(0, len(lista) - 1)
        
        if lista[indice_azar] == objetivo:
            print(f"¡Encontrado! El número {objetivo} estaba en el índice {indice_azar}.")
            print(f"Total de intentos realizados: {intentos}")
            encontrado = True

# Creamos una lista del 1 al 100 y buscamos el 7
mi_lista = list(range(1, 101))
algoritmo_las_vegas(mi_lista, 7)
```

```python
#Paralelismo - Procesamiento de datos
import multiprocessing
import time

def calcular_cuadrado(n):
    return n * n

if __name__ == "__main__":
    datos = list(range(1000000))
    
    # --- MODO SECUENCIAL ---
    inicio_seq = time.time()
    resultado_seq = [calcular_cuadrado(x) for x in datos]
    fin_seq = time.time()
    print(f"Tiempo Secuencial: {fin_seq - inicio_seq:.4f} segundos")

    # --- MODO PARALELO ---
    # Usamos Pool para repartir el trabajo entre los núcleos disponibles
    inicio_par = time.time()
    with multiprocessing.Pool() as pool:
        resultado_par = pool.map(calcular_cuadrado, datos)
    fin_par = time.time()
    print(f"Tiempo Paralelo: {fin_par - inicio_par:.4f} segundos")
    
    speedup = (fin_seq - inicio_seq) / (fin_par - inicio_par)
    print(f"Ganancia de velocidad (Speedup): {speedup:.2f}x")
```
