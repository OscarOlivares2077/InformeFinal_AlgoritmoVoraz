```python
def cambio_voraz(monto, monedas):
    monedas.sort(reverse=True) 
    resultado = []
    for moneda in monedas:
        while monto >= moneda:
            monto -= moneda
            resultado.append(moneda) 
            
    return resultado
sistema = [4, 3, 1]
objetivo = 6

print(f"Resultado Voraz: {cambio_voraz(objetivo, sistema)}")
```

```python
def fibonacci_recursivo(n):
    # Caso base: Si n es 0 o 1, el resultado es n
    if n <= 1:
        return n
    else:
        # Llamada recursiva: La función se llama a sí misma
        return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

# Prueba del algoritmo
n = 6
print(f"El término {n_termino} de Fibonacci es: {fibonacci_recursivo(n)}")
# Salida: 8
```
