```python
# ==============================================================
# Proyecto: Optimización de carga de bienes de ayuda humanitaria
# Institución de referencia: INDECI
# Algoritmo: Voraz
# ==============================================================
# Lista de bienes de ayuda humanitaria.
# Cada bien tiene nombre, peso, cantidad disponible y prioridad.
# Prioridad:
# 5 = Crítico
# 4 = Muy necesario
# 3 = Necesario
# 2 = Complementario
# 1 = Baja prioridad

bienes = [
    {"nombre": "Agua embotellada", "peso": 100, "cantidad": 4, "prioridad": 5},
    {"nombre": "Alimentos no perecibles", "peso": 120, "cantidad": 4, "prioridad": 5},
    {"nombre": "Carpas familiares", "peso": 150, "cantidad": 2, "prioridad": 4},
    {"nombre": "Frazadas", "peso": 80, "cantidad": 5, "prioridad": 4},
    {"nombre": "Kits de higiene", "peso": 60, "cantidad": 4, "prioridad": 3},
    {"nombre": "Colchonetas", "peso": 70, "cantidad": 4, "prioridad": 3},
    {"nombre": "Material de limpieza", "peso": 90, "cantidad": 2, "prioridad": 2},
    {"nombre": "Herramientas básicas", "peso": 200, "cantidad": 1, "prioridad": 2}
]

# Capacidad máxima del vehículo en kilogramos
capacidad_maxima = 1000


def carga_voraz_indeci(lista_bienes, capacidad):
    """
    Función que aplica un algoritmo voraz para seleccionar bienes
    de ayuda humanitaria sin superar la capacidad del vehículo.
    """

    # Se crea una copia de la lista para no modificar los datos originales.
    bienes_copia = []

    for bien in lista_bienes:
        bienes_copia.append(bien.copy())

    # Ordenamiento voraz:
    # Primero por mayor prioridad.
    # Si dos bienes tienen la misma prioridad, se elige el de menor peso.
    bienes_ordenados = sorted(
        bienes_copia,
        key=lambda x: (-x["prioridad"], x["peso"])
    )

    # Lista donde se almacenarán los bienes seleccionados.
    carga_seleccionada = []

    # Variables acumuladoras.
    peso_total = 0
    prioridad_total = 0

    # Recorrer cada bien ordenado según el criterio voraz.
    for bien in bienes_ordenados:

        # Mientras exista cantidad disponible del bien.
        while bien["cantidad"] > 0:

            # Verificar si el bien puede cargarse sin superar la capacidad.
            if peso_total + bien["peso"] <= capacidad:

                # Agregar el bien a la carga seleccionada.
                carga_seleccionada.append({
                    "nombre": bien["nombre"],
                    "peso": bien["peso"],
                    "prioridad": bien["prioridad"]
                })

                # Actualizar el peso total y la prioridad acumulada.
                peso_total += bien["peso"]
                prioridad_total += bien["prioridad"]

                # Disminuir la cantidad disponible del bien.
                bien["cantidad"] -= 1

            else:
                # Si el bien ya no entra, se pasa al siguiente producto.
                break

    # Calcular capacidad restante.
    capacidad_restante = capacidad - peso_total

    return carga_seleccionada, peso_total, capacidad_restante, prioridad_total


# Ejecutar el algoritmo
resultado, peso_total, capacidad_restante, prioridad_total = carga_voraz_indeci(
    bienes,
    capacidad_maxima
)

# Mostrar resultados
print("===== OPTIMIZACIÓN DE CARGA DE BIENES DE AYUDA HUMANITARIA - INDECI =====")
print("Capacidad máxima del vehículo:", capacidad_maxima, "kg")
print("Peso total cargado:", peso_total, "kg")
print("Capacidad restante:", capacidad_restante, "kg")
print("Prioridad total acumulada:", prioridad_total)

print("\nBienes seleccionados:")
for item in resultado:
    print(
        "-",
        item["nombre"],
        "| Peso:",
        item["peso"],
        "kg | Prioridad:",
        item["prioridad"]
    )
```

```python
# Pruebas con distintas capacidades de carga

capacidades_prueba = [500, 800, 1000, 1200, 1500]

print("===== ANÁLISIS EMPÍRICO =====")
print("Capacidad | Peso cargado | Restante | Prioridad total")
print("------------------------------------------------------")

for capacidad in capacidades_prueba:
    resultado, peso_total, restante, prioridad_total = carga_voraz_indeci(
        bienes,
        capacidad
    )

    print(
        capacidad,
        "kg |",
        peso_total,
        "kg |",
        restante,
        "kg |",
        prioridad_total
    )
```
